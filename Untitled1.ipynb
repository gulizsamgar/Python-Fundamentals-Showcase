car_crashes Veri Seti (Görev 1-3)

---

```python
# Gerekli kütüphaneleri import ediyoruz
import pandas as pd
import seaborn as sns

# Pandas gösterim ayarlarını yapılandırıyoruz
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 500)

# Görev 1: car_crashes veri setini yüklüyoruz
df = sns.load_dataset('car_crashes')

# Görev 2: Nümerik değişken isimlerini büyük harfe çevirip başına 'NUM_' eki ekliyoruz
numeric_cols = [
    'NUM_' + col.upper() if df[col].dtype != 'O' else col.upper()
    for col in df.columns
]
print(numeric_cols)

# Görev 3: İsmi 'no' içermeyen sütunların sonuna '_FLAG' eki ekliyoruz
flagged_cols = [
    col.upper() + '_FLAG' if 'no' not in col else col.upper()
    for col in df.columns
]
print(flagged_cols)

# Görev 4: 'abbrev' ve 'no_previous' dışındaki sütunlarla yeni DataFrame oluşturuyoruz
og_list = ['abbrev', 'no_previous']
new_cols = [col for col in df.columns if col not in og_list]
new_df = df[new_cols]
print(new_df.head())
```

Titanic Veri Seti (Görev 1-16)

```python


# Gerekli kütüphaneleri import ediyoruz
import numpy as np
import pandas as pd
import seaborn as sns

# Pandas gösterim ayarlarını yapılandırıyoruz
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Görev 1: Titanic veri setini yüklüyoruz
df = sns.load_dataset('titanic')

# Görev 2: Kadın ve erkek yolcu sayısını buluyoruz
print(df['sex'].value_counts())

# Görev 3: Her sütundaki benzersiz değer sayısını buluyoruz
print(df.nunique())

# Görev 4: 'pclass' sütununun benzersiz değerlerini alıyoruz
print(df['pclass'].unique())

# Görev 5: 'pclass' ve 'parch' sütunlarının benzersiz değer sayılarını buluyoruz
print(df[['pclass', 'parch']].nunique())

# Görev 6: 'embarked' sütununun veri tipini kontrol edip kategorik tipe dönüştürüyoruz
print(df['embarked'].dtype)
df['embarked'] = df['embarked'].astype('category')
print(df['embarked'].dtype)

# Görev 7: 'embarked' değeri 'C' olan kayıtları gösteriyoruz
print(df[df['embarked'] == 'C'].head(10))

# Görev 8: 'embarked' değeri 'S' olmayan kayıtları gösteriyoruz
print(df[df['embarked'] != 'S'].head(10))

# Görev 9: Yaşı 30'dan küçük ve kadın yolcuları seçiyoruz
print(df[(df['age'] < 30) & (df['sex'] == 'female')].head())

# Görev 10: 'fare' > 500 veya yaşı 70'ten büyük yolcuları seçiyoruz
print(df[(df['fare'] > 500) | (df['age'] > 70)].head())

# Görev 11: Her sütundaki boş değerlerin toplamını buluyoruz
print(df.isnull().sum())

# Görev 12: 'who' sütununu DataFrame'den çıkarıyoruz
df.drop('who', axis=1, inplace=True)

# Görev 13: 'deck' sütunundaki NaN değerleri en sık görülen değer (mode) ile dolduruyoruz
mode_deger = df['deck'].mode()[0]
df['deck'] = df['deck'].fillna(mode_deger)

# Görev 14: 'age' sütunundaki NaN değerleri medyan ile dolduruyoruz
medyan_deger = df['age'].median()
df['age'] = df['age'].fillna(medyan_deger)

# Görev 15: 'survived' değişkenini pclass ve sex kırılımında sum, count, mean ile analiz ediyoruz
print(df.groupby(['pclass', 'sex'])['survived'].agg(['sum', 'count', 'mean']))

# Görev 16: 30 yaş altındakilere 1, diğerlerine 0 atayan 'age_flag' sütunu oluşturuyoruz
def age_flag(age):
    return 1 if age < 30 else 0

df['age_flag'] = df['age'].apply(age_flag)
print(df.head())
```

Tips Veri Seti (Görev 17-23)

```python
# Gerekli kütüphaneleri import ediyoruz
import pandas as pd
import seaborn as sns

# Pandas gösterim ayarlarını yapılandırıyoruz
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Görev 17: Tips veri setini yüklüyoruz
df = sns.load_dataset('tips')

# Görev 18: 'time' kategorisine göre total_bill istatistiklerini (sum, min, max, mean) alıyoruz
print(df.groupby('time')['total_bill'].agg(['sum', 'min', 'max', 'mean']))

# Görev 19: 'time' ve 'day' bazlı total_bill istatistiklerini alıyoruz
print(df.groupby(['time', 'day'])['total_bill'].agg(['sum', 'min', 'max', 'mean']))

# Görev 20: Lunch zamanındaki kadın müşterilere göre total_bill ve tip istatistiklerini alıyoruz
filtered_df = df[(df['time'] == 'Lunch') & (df['sex'] == 'Female')]
print(filtered_df.groupby('day')[['total_bill', 'tip']].agg(['sum', 'min', 'max', 'mean']))

# Görev 21: size < 3 ve total_bill > 10 koşulundaki siparişlerin ortalamasını hesaplıyoruz
print(df.loc[(df['size'] < 3) & (df['total_bill'] > 10), 'total_bill'].mean())

# Görev 22: total_bill ve tip toplamını veren 'total_bill_tip_sum' sütununu oluşturuyoruz
df['total_bill_tip_sum'] = df['total_bill'] + df['tip']
print(df.head())

# Görev 23: 'total_bill_tip_sum'a göre sıralayıp en yüksek 30 kaydı seçiyoruz
top30_df = df.sort_values('total_bill_tip_sum', ascending=False).head(30)
print(top30_df.shape)
```
